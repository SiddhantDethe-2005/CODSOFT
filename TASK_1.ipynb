```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Titanic-Dataset.csv")
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data = data.drop(["Cabin"], axis = 1)
```

```python
data = data.drop(["PassengerId","Name","Ticket"], axis = 1)
```

```python
print(data.columns)
```

```python
data['Age'] = data['Age'].fillna(data['Age'].median())
```

```python
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
#Only Numbers
mapping_1 = {'male':1, 'female':0}
data['Sex'] = data['Sex'].map(mapping_1)
```

```python
mapping_2 = {'C':0, 'Q':1, 'S':2}
data['Embarked'] = data['Embarked'].map(mapping_2)
```

```python
print(data.head())
```

```python
sns.countplot(x = 'Survived', data = data)
plt.title("Survival Count")
plt.xticks([0,1],["Do not Survived", "Survived"])
plt.show
```

```python
sns.countplot(x = "Pclass", hue = "Survived", data = data)
plt.title("Survival by Class")
plt.legend(title = "Survived", labels = ['No', 'Yes'])
plt.show()
```

```python
sns.countplot(x = "Sex", hue = "Survived", data = data)
plt.title("Survival by Sex")
plt.legend(title = "Survived", labels = ['No', 'Yes'])
plt.show()
```

```python
sns.histplot(data['Age'], kde = True, bins = 30)
plt.title("Age Distribution of Passenger.")
plt.xlabel("Age")
plt.show()
```

```python
sns.histplot(x = 'Survived', y = 'Fare', data = data)
plt.title("Fare Paid vs Survival")
plt.xticks([0,1],["Not Survived", "Survived"])
plt.show()
```

```python
##Building machine Learning Model
```

```python
from sklearn.model_selection import train_test_split
x = data.drop('Survived', axis = 1)
y = data['Survived']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter = 1000)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=["Not Survived", "Survived"], yticklabels=["Not Survived", "Survived"])
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```
